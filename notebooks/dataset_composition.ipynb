```python
def get_tx_counts(db_name: str) -> dict:
    not_discarded = or_(
        GraphMappingBlockchain.discard_flag.is_(None),
        GraphMappingBlockchain.discard_flag != 1,
    )

    single_chain_total = (
        session.query(func.count(GraphMappingBlockchain.graph_id))
        .filter(
            GraphMappingBlockchain.bridge == db_name,
            GraphMappingBlockchain.cctx_graph_id.is_(None),
            not_discarded,
        )
        .scalar()
    )

    single_chain_anomalies = (
        session.query(func.count(GraphMappingBlockchain.graph_id))
        .filter(
            GraphMappingBlockchain.bridge == db_name,
            GraphMappingBlockchain.cctx_graph_id.is_(None),
            not_discarded,
            GraphMappingBlockchain.label == "anomaly",
        )
        .scalar()
    )

    cctx_total = (
        session.query(func.count(GraphMappingCrossChain.cctx_graph_id))
        .filter(GraphMappingCrossChain.bridge == db_name)
        .scalar()
    )

    cctx_anomalies = (
        session.query(func.count(GraphMappingCrossChain.cctx_graph_id))
        .filter(
            GraphMappingCrossChain.bridge == db_name,
            GraphMappingCrossChain.label.like("anomaly%"),
        )
        .scalar()
    )

    return {
        "single_chain_total": single_chain_total,
        "single_chain_anomalies": single_chain_anomalies,
        "cctx_total": cctx_total,
        "cctx_anomalies": cctx_anomalies,
    }


def format_sources(sources: list) -> str:
    return ", ".join(f"\\href{{{s['url']}}}{{{s['title']}}}" if "url" in s else s["title"] for s in sources)


def build_rows(incidents: list) -> list:
    rows = []
    for incident in incidents:
        counts = get_tx_counts(incident["db_name"])
        rows.append(
            {
                "Incident": incident["incident_name"],
                "Attack Date": incident["attack_date"],
                "Incident Type": incident["incident_type"],
                "Reason": incident["reason"],
                "Data Gathering Method": incident["gathering_method"],
                "#Single-chain tx's": counts["single_chain_total"],
                "of which anomalies (single-chain)": counts["single_chain_anomalies"],
                "#CCTX's": counts["cctx_total"],
                "of which anomalies (CCTX)": counts["cctx_anomalies"],
                "Sources": format_sources(incident["sources"]),
            }
        )
    return rows

def build_total_row(rows: list) -> dict:
    numeric_cols = [
        "#Single-chain tx's",
        "of which anomalies (single-chain)",
        "#CCTX's",
        "of which anomalies (CCTX)",
    ]
    total = {col: sum(row[col] for row in rows) for col in numeric_cols}
    total["Incident"] = f"Total ({len(rows)} incidents)"
    for col in ["Attack Date", "Incident Type", "Reason", "Data Gathering Method", "Sources"]:
        total[col] = ""
    return total
```

```python
def to_latex(
    rows: list,
    total_row: dict = None,
    caption: str = "Dataset composition per incident.",
    label: str = "table:evaluation-dataset-composition",
) -> str:
    columns = [
        "Incident",
        "Attack Date",
        "Incident Type",
        "Reason",
        "Data Gathering Method",
        "#Single-chain tx's",
        "of which anomalies (single-chain)",
        "#CCTX's",
        "of which anomalies (CCTX)",
        "Sources",
    ]
    header_labels = [
        "Incident",
        "Attack Date",
        "Incident Type",
        "Reason",
        "Method",
        "$\\#S$",
        "$\\#S_{A}$",
        "$\\#C$",
        "$\\#C_{A}$",
        "Sources",
    ]

    col_spec = "lllllrrrrl"

    lines = [
        "\\begin{table*}[t]",
        "\\centering",
        f"\\caption{{{caption}}}",
        f"\\label{{{label}}}",
        "\\resizebox{\\textwidth}{!}{%",
        f"\\begin{{tabular}}{{{col_spec}}}",
        "\\toprule",
        " & ".join(header_labels) + " \\\\",
        "\\midrule",
    ]
    for row in rows:
        lines.append(" & ".join(str(row[col]) for col in columns) + " \\\\")

    if total_row:
        lines.append("\\midrule")
        lines.append(" & ".join(
            f"\\textbf{{{total_row[col]}}}" if str(total_row[col]) != "" else ""
                for col in columns
            )
            + " \\\\"
        )
    lines += [
        "\\bottomrule",
        "\\end{tabular}%",
        "}",
        "\\end{table*}",
    ]
    return "\n".join(lines)

total_row = build_total_row(rows)
print(to_latex(rows, total_row=total_row))
```
